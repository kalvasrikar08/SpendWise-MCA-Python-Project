```python

# SPENDWISE - Personal Expense Management System

import sqlite3
import csv
from dataclasses import dataclass
from collections import defaultdict

# Database
db = sqlite3.connect("spendwise.db")
db.execute("""
CREATE TABLE IF NOT EXISTS expenses(
id INTEGER PRIMARY KEY AUTOINCREMENT,
title TEXT, category TEXT, amount REAL, date TEXT)
""")

# Data class
@dataclass
class Expense:
    title: str
    category: str
    amount: float
    date: str

# Add expense
def add_expense():
    title = input("Expense name: ")
    category = input("Category: ")
    amount = float(input("Amount: "))
    date = input("Date (YYYY-MM-DD): ")

    if amount <= 0:
        print("Amount must be positive!")
        return

    expense = Expense(title, category, amount, date)

    db.execute(
        "INSERT INTO expenses(title,category,amount,date) VALUES(?,?,?,?)",
        (expense.title, expense.category, expense.amount, expense.date)
    )
    db.commit()
    print("Expense added successfully!")

# View expenses
def view_expenses():
    records = db.execute("SELECT * FROM expenses").fetchall()

    for record in records:
        print(record)

# Expense summary
def summary():
    records = db.execute(
        "SELECT category, SUM(amount) FROM expenses GROUP BY category"
    ).fetchall()

    print("\nExpense Summary")
    for category, amount in records:
        print(category, ":", amount)

# Monthly spending prediction
def predict_expense():
    records = db.execute(
        "SELECT date, SUM(amount) FROM expenses GROUP BY date"
    ).fetchall()

    if not records:
        print("No expense data available!")
        return

    average = sum(row[1] for row in records) / len(records)
    print("Estimated 30-day spending:", round(average * 30, 2))

# Export CSV
def export_report():
    records = db.execute("SELECT * FROM expenses").fetchall()

    with open("spendwise_report.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["ID", "Title", "Category", "Amount", "Date"])
        writer.writerows(records)

    print("CSV report exported!")

# Main menu
while True:
    print("""
===== SPENDWISE =====
1. Add Expense
2. View Expenses
3. Expense Summary
4. Predict Monthly Spending
5. Export CSV
6. Exit
""")

    choice = input("Enter choice: ")

    if choice == "1":
        add_expense()
    elif choice == "2":
        view_expenses()
    elif choice == "3":
        summary()
    elif choice == "4":
        predict_expense()
    elif choice == "5":
        export_report()
    elif choice == "6":
        db.close()
        print("Thank you for using SpendWise!")
        break
    else:
        print("Invalid choice!")
```

```python

# SPENDWISE - Advanced OOP Module

from dataclasses import dataclass

# Parent class
class Transaction:

    def __init__(self, title, amount, category):
        self.title = title
        self.amount = amount
        self.category = category

    def display(self):
        return f"{self.title} - Rs.{self.amount}"

    def __str__(self):
        return self.display()


# Child class: Expense
class Expense(Transaction):

    def __init__(self, title, amount, category):
        super().__init__(title, amount, category)
        self._amount = amount

    @property
    def amount(self):
        return self._amount

    @amount.setter
    def amount(self, value):
        if value <= 0:
            raise ValueError("Amount must be positive!")
        self._amount = value

    def display(self):
        return f"Expense: {self.title} | Rs.{self.amount} | {self.category}"


# Child class: Income
class Income(Transaction):

    def display(self):
        return f"Income: {self.title} | Rs.{self.amount}"


# Dataclass
@dataclass
class Budget:
    category: str
    limit: float


# Composition
class FinanceManager:

    def __init__(self):
        self.transactions = []
        self.budgets = []

    def add_transaction(self, transaction):
        self.transactions.append(transaction)

    def add_budget(self, budget):
        self.budgets.append(budget)

    def show_transactions(self):
        for transaction in self.transactions:
            print(transaction.display())

    def total_expenses(self):
        return sum(
            t.amount for t in self.transactions
            if isinstance(t, Expense)
        )

    def total_income(self):
        return sum(
            t.amount for t in self.transactions
            if isinstance(t, Income)
        )


# Main program
manager = FinanceManager()

manager.add_transaction(Expense("Food", 500, "Food"))
manager.add_transaction(Expense("Travel", 300, "Transport"))
manager.add_transaction(Income("Salary", 15000, "Job"))

manager.add_budget(Budget("Food", 3000))

print("===== SPENDWISE OOP SYSTEM =====")

manager.show_transactions()

print("\nTotal Expenses:", manager.total_expenses())
print("Total Income:", manager.total_income())

print("\nBudgets:")
for budget in manager.budgets:
    print(budget.category, ":", budget.limit)

print("\nRemaining Balance:",
      manager.total_income() - manager.total_expenses())
```

```python

%%writefile /content/SpendWise/modules/advanced.py

# List comprehension
def high_expenses(expenses):
    return [e for e in expenses if e["amount"] > 500]

# Dictionary comprehension
def expense_summary(expenses):
    return {
        e["category"]: e["amount"]
        for e in expenses
    }

# Set comprehension
def categories(expenses):
    return {e["category"] for e in expenses}

# Generator function
def expense_generator(expenses):
    for expense in expenses:
        yield expense["amount"]

# Generator expression
def total_expense(expenses):
    return sum(e["amount"] for e in expenses)
```

```python

%%writefile /content/SpendWise/modules/advanced_tools.py

from functools import wraps

# 1. Custom Iterator
class ExpenseIterator:

    def __init__(self, expenses):
        self.expenses = expenses
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.expenses):
            raise StopIteration

        value = self.expenses[self.index]
        self.index += 1
        return value


# 2. Function Decorator
def track_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Executing:", func.__name__)
        result = func(*args, **kwargs)
        print("Execution completed!")
        return result
    return wrapper


# 3. Context Manager
class ExpenseFile:

    def __enter__(self):
        print("Opening expense file")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing expense file")
        return False


# Decorator example
@track_time
def calculate_total(expenses):
    return sum(expenses)
```

```python

%%writefile /content/SpendWise/modules/quality.py

from typing import List, Dict

# Type hints
def calculate_total(expenses: List[float]) -> float:
    return sum(expenses)


def calculate_average(expenses: List[float]) -> float:
    if not expenses:
        return 0.0
    return calculate_total(expenses) / len(expenses)


def category_count(categories: List[str]) -> Dict[str, int]:
    result = {}

    for category in categories:
        result[category] = result.get(category, 0) + 1

    return result


# Main function
def main():
    expenses: List[float] = [200, 500, 800, 300]
    categories: List[str] = ["Food", "Travel", "Food", "Shopping"]

    print("Total:", calculate_total(expenses))
    print("Average:", calculate_average(expenses))
    print("Category Count:", category_count(categories))


if __name__ == "__main__":
    main()
```
